# Split data
- randomly select 10 data points as demo data

```python
import pandas as pd

full_df = pd.read_csv('../data/data.csv')

# Separate the dataset by class
m_df = full_df[full_df['diagnosis'] == 'M']
b_df = full_df[full_df['diagnosis'] == 'B']

# Randomly sample exactly 5 from each class
demo_m = m_df.sample(n=5, random_state=18)
demo_b = b_df.sample(n=5, random_state=18)

# Combine into a 10-patient demo set and shuffle them 
demo_df = pd.concat([demo_b, demo_m])

# Drop these 10 specific patients from the main dataset
train_df = full_df.drop(index=demo_df.index)

# Save the clean files
demo_df.to_csv('../data/demo_data.csv', index=False)
train_df.to_csv('../data/train_data.csv', index=False)

# Verify the math
print(f"Total original patients: {len(full_df)}")
print(f"Demo patients saved: {len(demo_df)} (Perfectly balanced: 5 M, 5 B)")
print(f"Remaining data for your training script: {len(train_df)}")
```
